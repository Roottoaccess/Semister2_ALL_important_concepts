```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('titanic.csv')
df.head()
```

```python
df.shape
```

```python
df.isna().sum()
```

```python
# df.drop(df['Cabin'],axis=1, inplace=True)
del df['Cabin']
df.head()
```

```python
x = ['Sex','Embarked']
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
for i in x:
    df[i] = le.fit_transform(df[i])
```

```python
df.head()
```

```python
df.isna().sum()
```

```python
df['Age'].fillna(df['Age'].mean(),inplace=True)
df.head()
```

```python
df.isna().sum()
```

```python
df.head()
```

```python
df.corr(numeric_only=True)
```

```python
x = df[['Pclass', 'Sex', 'Age', 'SibSp', 'Parch', 'Fare', 'Embarked']]
y = df.Survived
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42,shuffle=True)
```

```python
# Building model Decision tree classifier -> because its a binary classification
from sklearn.tree import DecisionTreeClassifier
dtc = DecisionTreeClassifier()
dtc.fit(x_train,y_train)
```

```python
dtc.score(x_test,y_test)
```

```python
dtc.score(x_train,y_train)
```

```python
y_pred = dtc.predict(x_test)
y_pred
```

```python
from sklearn.metrics import confusion_matrix
confusion_matrix(y_test,y_pred)
```

```python
# Implementing the logictic regression -> Model number 2
from sklearn.linear_model import LogisticRegression
lor = LogisticRegression()
lor.fit(x_train,y_train)
```

```python
lor.score(x_test,y_test)
```

```python
lor.score(x_train,y_train)
```

```python
# Implementing the Random forest classifier for removing the overfitting problem in the Decision tree classifier
from sklearn.ensemble import RandomForestClassifier
rfc = RandomForestClassifier(n_estimators=100,max_depth=5,random_state=42) # What is this ?
rfc.fit(x_train,y_train)
```

```python
rfc.score(x_test,y_test)
```

```python
rfc.score(x_train,y_train)
```

# GridSearch

```python
from sklearn.model_selection import GridSearchCV
from sklearn.svm import SVC
svm = SVC()
gv = {
    "kernel":["rbf","sigmoid"],
    "C": [1000,100,10,1],
    "gamma": ["auto","scale"]
}
gs = GridSearchCV(estimator = svm,
                  param_grid=gv,
                  scoring="accuracy",
                  cv=5, verbose=1)

gs.fit(x_train,y_train)
```

```python

gs.best_estimator_
```

```python
gs.score(x_train,y_train)
```

```python
gs.score(x_test,y_test)
```
